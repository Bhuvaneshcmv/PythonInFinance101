# Pt.36) Basic technical Indicators

```python
asianPaintsDf = pd.read_csv('AsianPaints.csv')
asianPaintsDf.columns = asianPaintsDf.columns.str.strip().str.replace(' ','_')
asianPaintsDf['DATE'] = pd.to_datetime(asianPaintsDf['DATE'])
asianPaintsDf = asianPaintsDf.sort_values('DATE')
asianPaintsDf.set_index('DATE', inplace=True)
columns_to_convert = ['OPEN','HIGH','LOW','CLOSE']
for col in columns_to_convert:
    asianPaintsDf[col] = asianPaintsDf[col].str.replace(',','').astype(float)
asianPaintsDfColumnsToConvert = asianPaintsDf[['OPEN','HIGH','LOW','CLOSE']]
print(asianPaintsDf)
```

## Candle Stick

```python
candleStickAsianPaintsDf = asianPaintsDf[['OPEN','HIGH','LOW','CLOSE']].copy()
candleStickAsianPaintsDf.columns = ['Open','High','Low','Close']
mpf.plot(candleStickAsianPaintsDf, type='candle', style='charles', title='Asian Paints - Candle Stick Chart', ylabel ='Price(INR)', volume=False)
```

## Simple moving averages

```python
asianPaintsDf['SMA20'] = asianPaintsDf['CLOSE'].rolling(window=20).mean()
asianPaintsDf['SMA50'] = asianPaintsDf['CLOSE'].rolling(window=50).mean()
plt.figure(figsize=(12,6))
plt.plot(asianPaintsDf['CLOSE'], label='Close Price', color='black')
plt.plot(asianPaintsDf['SMA20'], label='SMA 20', color='green')
plt.plot(asianPaintsDf['SMA50'], label='SMA 50', color='red')
plt.title('Asian Paints - Simple Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price(INR)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

## Relative Strength Index (RSI)

```python
delta = asianPaintsDf['CLOSE'].diff()
gain = delta.where(delta>0, 0)
loss = delta.where(delta<0, 0)

avg_gain = gain.rolling(window=14).mean()
avg_loss = -loss.rolling(window=14).mean()

rs = avg_gain/avg_loss
asianPaintsDf['RSI'] = 100 - (100/(1+rs))

plt.figure(figsize=(10,4))
plt.plot(asianPaintsDf['RSI'], label='RSI', color = 'purple')
plt.axhline(70, linestyle='--', color='red', label ='Overbought')
plt.axhline(30, linestyle='--', color='green', label ='OverSold')
plt.title('Asian Paints - RSI (14-day')
plt.xlabel('Date')
plt.ylabel('RSI Value')
plt.legend()
plt.tight_layout()
plt.show()
```

## Moving Average Convergence Divergence (MACD)

## Portfolio Optimization

```python
#Download Stock data
stocks = ['AAPL','MSFT','GOOGL','AMZN']
stocks_data = yf.download(stocks, start ='2020-01-01', end='2025-08-15')['Close']
```

```python
#Calculate log returns
returns = np.log(stocks_data/stocks_data.shift(1)).dropna()
```

```python
mean_stock_returns = returns.mean()
stock_cov_matrix = returns.cov()
```

```python
def portfolio_performance(weights, mean_returns, cov_matrix):
    returns = np.dot(weights, mean_returns)*252
    risk = np.sqrt(np.dot(weights.T,np.dot(cov_matrix, weights))*252)
    return risk, returns
```

```python
def neg_sharpe_ratio(weights, mean_returns, cov_matrix, risk_free_rate=0.02):
    risk,ret = portfolio_performance(weights, mean_returns, cov_matrix)
    return  -(ret - risk_free_rate)/risk
```

```python
constraints = ({'type': 'eq' , 'fun' : lambda x: np.sum(x)-1})
bounds = tuple((0,1) for asset in range(len(stocks)))
```

```python
init_guess = len(stocks)*[1. / len(stocks)]

opt_results = minimize(neg_sharpe_ratio, init_guess,
                       args=(mean_stock_returns, stock_cov_matrix),
                       method='SLSQP', bounds = bounds, constraints= constraints)

optimal_weights = opt_results.x
print("Optimal portfolio weights ", dict(zip(stocks,optimal_weights)))
```

```python
def efficient_frontier(mean_returns, cov_matrix, returns_range):
    target_returns = np.linspace(returns_range[0],returns_range[1],50)
    risks = []
    rets = []
    for r in target_returns:
        constraints = ({'type' : 'eq' , 'fun' : lambda x:np.sum(x)-1},
                       {'type' : 'eq' , 'fun' : lambda x:portfolio_performance(x, mean_returns, cov_matrix)[1] -r})
        result = minimize(lambda w: portfolio_performance(w, mean_returns, cov_matrix)[0],
                          init_guess, method='SLSQP', bounds = bounds, constraints = constraints)
        if result.success:
            risks.append(result.fun)
            rets.append(r)
    return risks, rets
risks, rets = efficient_frontier(mean_stock_returns, stock_cov_matrix, (0.1,0.3))
```

```python
plt.figure(figsize = (12,8))
plt.plot(risks, rets, 'b--', label= 'EFFICIENT Frontier')
plt.xlabel('RISK (Volatility)')
plt.ylabel('Expected Return')
plt.title('Efficient Frontier - Markowitz Portfolio Optimization')
plt.legend()
plt.show()
```

## Markowitz model demonstration

```python
TRADING_DAYS = 252
PORTFOLIOS = 100000
start_date = '2013-09-01'
end_date = '2025-08-15'
stock = 'GOOG', 'MSFT', 'AAPL', 'AMZN', 'NVDA', 'AMD', 'TSLA'
```

```python
def download_data(start_date, end_date, stocks):
    close_data = {}
    for s in stocks:
        ticker = yf.Ticker(s)
        close_data[s] = ticker.history(start = start_date, end = end_date)['Close']
    return pd.DataFrame(close_data)
```

```python
def compute_returns(data):
    log_return = np.log(data / data.shift(1))
    return log_return[1:]
```

```python
def compute_statistics(returns):
    print('ANNUAL MEAN \n '+str(returns.mean() * TRADING_DAYS))
    print('\n COVARIANCE MEAN: \n '+str(returns.cov()*TRADING_DAYS))
```

```python
def compute_mean_variance(returns,weights):
    ret = np.sum(returns.mean()*weights)*TRADING_DAYS
    vol = np.sqrt(np.dot(weights.T,np.dot(returns.cov()*TRADING_DAYS,weights)))
    print('EXPECTED PORTFOLIO MEAN: '+str(ret))
    print('Expected portfolio volatility: '+str(vol))
```

```python
def generate_portfolios(returns,stocks):
    means = []
    risks = []
    weights = []

    for _ in range(PORTFOLIOS):
        weight = np.random.random(len(stocks))
        weight /= np.sum(weight)
        mean = np.sum(returns.mean() * weight) * TRADING_DAYS
        risk = np.sqrt(np.dot(weight.T, np.dot(returns.cov() * TRADING_DAYS, weight)))
        weights.append(weight)
        means.append(mean)
        risks.append(risk)
    return np.array(weights), np.array(means), np.array(risks)
```

```python
# Portfolio statistics, containing returns, volatility and sharpe ratio
def portfolio_statistics(weights, returns):
    ret = np.sum(returns.mean() * weights) * TRADING_DAYS
    vol = np.sqrt(np.dot(weights.T, np.dot(returns.cov() * TRADING_DAYS, weights)))
    return np.array([ret, vol, ret/vol])


def minimize_sharpe_ratio(weights, returns):
    return -portfolio_statistics(weights, returns)[2]
```

```python
# Minimize f(x) = 0, where sum of weights equivalent to 1
def optimize_portfolio(weights, returns, stocks):
    constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
    bounds = tuple((0, 1) for _ in range(len(stocks)))

    optimum = optimization.minimize(
        fun=minimize_sharpe_ratio,
        x0=weights[0],
        args=returns,
        method='SLSQP',
        bounds=bounds,
        constraints=constraints
    )

    print('Optimal portfolio: ' + str(optimum['x'].round(3)))
    print(
        'Expected return, volatility and Sharpe ratio: ' +
        str(portfolio_statistics(optimum['x'].round(3), returns))
    )

    return optimum
```

```python
# Plot data
def plot_data(data):
    data.plot(figsize=(10, 6))
    plt.show()


# Plot portfolios
def plot_portfolios(returns, vol):
    plt.figure(figsize=(10, 6))
    plt.grid(True)

    plt.scatter(
        vol,
        returns,
        c=returns / vol,
        marker='o'
    )

    plt.xlabel('Expected Volatility')
    plt.ylabel('Expected Return')
    plt.colorbar(label='Sharpe Ratio')
    plt.show()


# Plot optimal portfolio
def plot_optimal_portfolio(opt, returns, portfolio_returns, vol):
    plt.figure(figsize=(10, 6))
    plt.grid(True)

    plt.scatter(
        vol,
        portfolio_returns,
        c=portfolio_returns / vol,
        marker='o'
    )

    plt.xlabel('Expected Volatility')
    plt.ylabel('Expected Return')
    plt.colorbar(label='Sharpe Ratio')

    plt.plot(
        portfolio_statistics(opt['x'], returns)[1],
        portfolio_statistics(opt['x'], returns)[0],
        'g*',
        markersize=20.0
    )

    plt.show()


stocks = ['GOOG', 'MSFT', 'AAPL', 'AMZN', 'NVDA', 'AMD', 'TSLA']

start_date = '2013-09-01'
end_date = '2025-08-15'

data = download_data(start_date, end_date, stocks)
plot_data(data)
```

```python
returns = compute_returns(data)
compute_statistics(returns)
```

```python
weights,means,risks = generate_portfolios(returns, stocks)
plot_portfolios(means,risks)
```

```python
optimum = optimize_portfolio(weights, returns, stocks)
plot_optimal_portfolio(optimum,returns, means,risks)
```

```python
df_merged = pd.merge(df_tcs_subset, df_infy_subset, on="DATE", suffixes=("_TCS", "_INFY"))

print(df_merged)
```

```python
# Daily returns
df_merged["RET_TCS"] = df_merged["CLOSE_TCS"].pct_change()
df_merged["RET_INFY"] = df_merged["CLOSE_INFY"].pct_change()
# .pct_change() is a pandas method that calculates the percentage change between the current row and the previous row.
# Return = (Today’s Price – Yesterday’s Price)/Yesterday’s Price
df_merged = df_merged.dropna()
# .dropna() removes all rows that contain any NaN (missing) values
df_merged.head()
```

```python
monthly = df_merged.set_index("DATE").resample("ME").last()
# df_merged.set_index("Date")
# Takes the column "Date" and makes it the index of the DataFrame
# This is required because resampling works only with a Datetime index
# .resample("M")
# Resampling means converting data from one frequency to another.
# "M" = Month-end frequency.
# .last()
# From each monthly bucket, it takes the last available row (closing price at month-end).
# Useful in finance because investors usually track month-end values for returns
monthly["RET_TCS"] = monthly["CLOSE_TCS"].pct_change()
monthly["RET_INFY"] = monthly["CLOSE_INFY"].pct_change()
monthly = monthly.dropna()
monthly.head()
```

```python
mean_returns = monthly[["RET_TCS", "RET_INFY"]].mean()
volatility = monthly[["RET_TCS", "RET_INFY"]].std()
print("TCS - Expected Return:", round(mean_returns.RET_TCS*100,2), "%, Risk:", round(volatility.RET_TCS*100,2), "%")
print("INFY - Expected Return:", round(mean_returns.RET_INFY*100,2), "%, Risk:", round(volatility.RET_INFY*100,2), "%")
```

```python
risk_free = 0.06/12 # monthly
excess_returns = mean_returns - risk_free
print("Excess Return - TCS:", round(excess_returns.RET_TCS*100,2), "%")
print("Excess Return - INFY:", round(excess_returns.RET_INFY*100,2), "%")
```

```python
#warning since there is not enough data to compute the covariance and variance correctly.
w_TCS, w_INFY = 0.5, 0.5
# Portfolio expected return
port_return = w_TCS*mean_returns.RET_TCS + w_INFY*mean_returns.RET_INFY
# Portfolio risk
cov_matrix = monthly[["RET_TCS", "RET_INFY"]].cov()
port_risk = (w_TCS**2 * monthly.RET_TCS.var() +
 w_INFY**2 * monthly.RET_INFY.var() +
 2*w_TCS*w_INFY*cov_matrix.loc["RET_TCS","RET_INFY"])**0.5
# Covariance matrix shows how TCS and Infosys returns move together
print("Portfolio Expected Return:", round(port_return*100,2), "%")
print("Portfolio Risk (Std Dev):", round(port_risk*100,2), "%")
# round(port_return*100,2) - converts decimal to %age form upto 2 decimal points
```

```python
# Wealth index (base 100)
monthly["Wealth_TCS"] = (1 + monthly["RET_TCS"]).cumprod() * 100
# This code creates a Wealth Index for TCS that starts at 100 and tracks
 # how an investment of ₹100 would grow over time based on monthly returns
# (1 + monthly["ret_TCS"]) Converts returns into growth factors
# .cumprod() Takes the cumulative product of these growth factors over time.
# Running peak
monthly["Peak_TCS"] = monthly["Wealth_TCS"].cummax()
# .cummax() cummax() gives the cumulative maximum value seen so far
# Drawdown
monthly["Drawdown_TCS"] = (monthly["Wealth_TCS"] - monthly["Peak_TCS"]) / monthly["Peak_TCS"]
# Drawdown measures how much the investment has fallen from its highest value so far
# (Wealth_TCS - Peak_TCS) / Peak_TCS converts the difference into a percentage drop from peak
max_dd_tcs = monthly["Drawdown_TCS"].min()
print("Maximum Drawdown TCS:", round(max_dd_tcs*100,2), "%")
# The Maximum Drawdown is the worst decline from a peak in the wealth index
 # during the entire period
# It tells investors the maximum downside risk they would have faced
# monthly["Drawdown_TCS"] contains the drawdown values (negative percentages)
# min() finds the lowest (most negative) drawdown
# That represents the largest drop from peak wealth
# Wealth index (base 100)
monthly["Wealth_INFY"] = (1 + monthly["RET_INFY"]).cumprod() * 100
# Running peak
monthly["Peak_INFY"] = monthly["Wealth_INFY"].cummax()
# Drawdown
monthly["Drawdown_INFY"] = (monthly["Wealth_INFY"] - monthly["Peak_INFY"]) / monthly["Peak_INFY"]
max_dd_infy = monthly["Drawdown_INFY"].min()
print("Maximum Drawdown INFY:", round(max_dd_infy*100,2), "%")
```

## Loan and EMI calculation

```python
principal = float(input('Enter the loan amount'))
annual_rate = float(input('Enter the annual rate of interest'))
tenure_years = float(input('Enter the loan tenure in years'))

monthly_rate = annual_rate / 12 /100
tenure_months = tenure_years * 12


```

```python
emi = (principal * monthly_rate * (1 + monthly_rate)**tenure_months)/((1 + monthly_rate)**tenure_months -1)

print(emi)

print(f'Monthly EMi : Rs.{round(emi,2)}')
print(f'Total Payment : Rs.{round(emi * tenure_months,2)} ')
print(f'Total Interest : Rs.{round(emi * tenure_months - principal,2)} ')
```

```python
months = list(range(1, int(tenure_months) + 1))

principal_paid = []
interest_paid = []
remaining_principal = principal

for month in months:
    interest = remaining_principal * monthly_rate
    principal_component = emi - interest
    principal_paid.append(principal_component)
    interest_paid.append(interest)
    remaining_principal -= principal_component

plt.figure(figsize=(10,5))
plt.stackplot(months, interest_paid, principal_paid, colors=['red', 'green'] , labels = ['Interest', 'principal'])
plt.title('EMI breakdown over time')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Using Numpy_financials

p = 100000
annual_rate = 9.5
tenure_years = 15

n = tenure_years * 12
r = annual_rate / 12 / 100

emi = -npf.pmt(r,n,p)

print(f'Monthly EMI : {round(emi,2)}')
print(f'Tenure Months : {n}')

```

```python
schedule = []
balance = p

for m in range(1, n+1):
    interest = balance*r
    principal_comp = emi - interest
    balance = max(0.0, balance - principal_comp)
    schedule.append([m, emi, interest, principal_comp ,balance])

df = pd.DataFrame(schedule,
                  columns = ['Month', 'EMI', 'Interest', 'Principal', 'Balance'])
df.head(12)
```
